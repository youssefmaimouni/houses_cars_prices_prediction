Importation de la bibliothèque Pandas

```python
import pandas as pd
```

Charger les données depuis un fichier CSV sur une DataFrame

```python
df = pd.read_csv('avito_car_dataset_ALL.csv')
```

Supprimer les colonnes non nécessaires

```python
df.drop(['Unnamed: 0','Lien','Ville' ,	'Secteur'],axis=1,inplace=True)
```

Voir les données manquantes

```python
df.isnull().sum()
```

Remplacer les données manquantes dans la colonne 'Nombre de portes' par la moyenne des valeurs existantes

```python
df['Nombre de portes'].fillna(df['Nombre de portes'].mode()[0],inplace=True)
df['Nombre de portes']=df['Nombre de portes'].astype(int)
df
```

```python
not_complete_columns = ['Origine','Premiére main','état']
for c in not_complete_columns:
    df[c].fillna(df[c].mode()[0], inplace=True)
df
```

```python
df.info()
```

Voir les données manquantes une autre fois pour vérifier qu'il n'y a aucune donnée manquante

```python
df.isnull().sum()
```

Voir les statistiques de la DataFrame

```python
df.describe()
```

```python
df.head()
```

Transformer le kilométrage de la gamme en point médian

```python
df['Kilométrage'] = df['Kilométrage'].str.replace(' ', '')
df[['Kilométrage_Min', 'Kilométrage_Max']] = df['Kilométrage'].str.extract('(\d+)-(\d+)', expand=True).astype(int)
df['Kilométrage']=(df['Kilométrage_Min']+df['Kilométrage_Max']+1)/2
df['Kilométrage']=df['Kilométrage'].astype(int)
df = df.drop(['Kilométrage_Min', 'Kilométrage_Max'], axis=1)
```

```python
df
```

Unification des données : transformer toutes les valeurs en minuscules et enlever les accents

```python
string_columns=['Modéle','Marque','Type de carburant','Boite de vitesses','Origine','Premiére main','état']
for column in string_columns:
    df[column] = df[column].str.lower().str.strip()
    df[column] = df[column].str.replace('é','e')

df
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Boite de vitesses'].unique()
```

Remplacer les valeurs manquantes dans la colonne 'Boîte de vitesses' par la mode de la colonne

```python
df['Boite de vitesses'] = df['Boite de vitesses'].replace('--', df['Boite de vitesses'].mode()[0])
df
```

Extraire le nombre de valeurs uniques par colonne

```python
def get_unique_values_count(data):
    """Retourne un dictionnaire avec le nombre de valeurs uniques par colonne."""
    return {col: data[col].nunique() for col in data.columns}
unique_values_count=get_unique_values_count(df)
unique_values_count
```

Supprimer les colonnes où il y a une valeur uniques

```python
df.drop(['Jantes aluminium','Systéme de navigation/GPS','Toit ouvrant' ,'Siéges cuir','Radar de recul', 'Caméra de recul', 'Vitres électriques','Régulateur de vitesse', 'Limiteur de vitesse', 'Ordinateur de bord', 'Verrouillage centralisé é distance'],axis=1,inplace=True)
df
```

Transformer la colonne 'Première main' en booléen

```python
df['Premiére main']=df['Premiére main']=='oui'
df
```

Transformer les colonnes contenant des valeurs booléennes en entiers

```python
bool_columns = ['Airbags','Climatisation','ABS','ESP','CD/MP3/Bluetooth','Premiére main']
for c in bool_columns:
    df[c]=df[c].astype(int)

df
```

Transformer les titres des colonnes en minuscules, remplacer les espaces par des underscores et enlever les accents

```python
df.columns=df.columns.str.lower().str.strip()
df.columns=df.columns.str.replace(' ','_')
df.columns=df.columns.str.replace('é','e')
df.columns
```

```python
df
```

Charger la DataFrame après nettoyage dans un fichier CSV

```python
Q1 = df['prix'].quantile(0.25)
Q3 = df['prix'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
df=df[(df['prix'] >= lower_bound) & (df['prix'] <= upper_bound)]
```

```python
df.to_csv('../../cleaned_data/cars_data_cleaned.csv',index=False)
```
