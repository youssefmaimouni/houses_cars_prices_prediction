# Houses Cleaning notebook

```python
import pandas as pd
import numpy as np
```

```python
# Load the uploaded file to inspect its structure
file_path = '../../casablanca_houses_data.csv'
df = pd.read_csv(file_path)
df
```

```python
df.isnull().sum()
```

```python
#clean type column to keep only the first word
df['Type'] = df['Type'].apply(lambda x: str(x).split(' ')[0])
df
```

```python
#remove rows with missing price and clean price (remove DH and change type)
# Clean the 'Price' column
df['Price'] = df['Price'].str.replace('DH', '').str.replace(' ', '').astype(str)
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

# Rename the column to 'Price in DH'
df.rename(columns={'Price': 'Price in DH'}, inplace=True)
df
```

```python
df.describe()
```

```python
#convert tags from string to list
df['Tags'] = df.Tags.apply(eval)
df
```

```python
df
```

```python
#clean area, rooms, bedrooms, bathrooms and floor columns
df['Area'] = df['Area'].str.replace(r'\D+','').astype('float')
df['Rooms'] = df['Rooms'].str.replace(r'\D+','').astype('int')
df['Bedrooms'] = df['Bedrooms'].str.replace(r'\D+','').astype('int')
df['Bathrooms'] = df['Bathrooms'].str.replace(r'\D+','').astype('int')
df['Floor'] = df['Floor'].str.replace(r'\D+','').astype('int')
df
```
