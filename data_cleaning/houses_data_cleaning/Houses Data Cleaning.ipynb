# Houses Cleaning notebook

```python
import pandas as pd
```

```python
# Load the uploaded file to inspect its structure
file_path = '../../casablanca_houses_data.csv'
data = pd.read_csv(file_path)
data
```

```python
# Display basic information of the dataset
data_info = data.info()
```

```python
data.isnull().sum()
```

```python
data.describe()
```

```python
# Rename the 'Price' column to 'Price in DH' after cleaning
# Clean the 'Price' column
data['Price'] = data['Price'].str.replace('DH', '').str.replace(' ', '').astype(str)
data['Price'] = pd.to_numeric(data['Price'], errors='coerce')

# Rename the column to 'Price in DH'
data.rename(columns={'Price': 'Price in DH'}, inplace=True)
data
```

```python
data.describe()
```

```python
# Fill missing numeric values with the mean
numeric_columns = ['Price in DH', 'Latitude', 'Longitude']
for col in numeric_columns:
    if col in data.columns:
        data[col].fillna(data[col].mean(), inplace=True)

data
```

```python
# Fill missing non-numeric values with the mode
non_numeric_columns = ['Type']
for col in non_numeric_columns:
    if col in data.columns:
        data[col].fillna(data[col].mode()[0], inplace=True)

data
```
