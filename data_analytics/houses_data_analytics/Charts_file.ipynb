# Charts
>These six charts should provide a comprehensive overview of your dataset

- Distribution of property types (Bar Chart).
- Distribution of prices (Histogram).
- Price vs. Area (Scatter Plot).
- Property condition proportions (Pie Chart).
- Price distribution by house age (Box Plot).
- Property distribution by location with price variations (Geospatial Plot).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the uploaded dataset to inspect its structure
file_path = '../../data_cleaning/houses_data_cleaning/Cleaned_dataset.csv'
dataset = pd.read_csv(file_path)

# Display the first few rows and basic information of the dataset
dataset.head(), dataset.info()
```

```python
# Set a theme for better visuals
sns.set_theme(style="whitegrid")

# 1. Bar Chart: Distribution of Property Types
plt.figure(figsize=(10, 6))
sns.countplot(y="Type", data=dataset, order=dataset["Type"].value_counts().index)
plt.title("Distribution of Property Types")
plt.xlabel("Count")
plt.ylabel("Property Type")
plt.tight_layout()
plt.show()
```

```python
# 2. Histogram: Distribution of Prices
plt.figure(figsize=(10, 6))
sns.histplot(dataset["Price in DH"], bins=30, kde=True)
plt.title("Distribution of Property Prices")
plt.xlabel("Price in DH")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# 3. Scatter Plot: Price vs. Area
plt.figure(figsize=(10, 6))
sns.scatterplot(x="Area in m²", y="Price in DH", data=dataset, alpha=0.6)
plt.title("Price vs. Area")
plt.xlabel("Area in m²")
plt.ylabel("Price in DH")
plt.tight_layout()
plt.show()
```

```python
# 4. Pie Chart: Property Condition Proportion
condition_counts = dataset["Condition"].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(condition_counts, labels=condition_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette("pastel"))
plt.title("Property Condition Proportion")
plt.tight_layout()
plt.show()
```

```python
# 5. Box Plot: Price by House Age
plt.figure(figsize=(10, 6))
sns.boxplot(x="House Age in ans", y="Price in DH", data=dataset)
plt.title("Price Distribution by House Age")
plt.xlabel("House Age (in years)")
plt.ylabel("Price in DH")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Adjust Geospatial Plot: Use scatterplot with color based on Price in DH
plt.figure(figsize=(10, 6))
scatter = plt.scatter(x=dataset["Longitude"], y=dataset["Latitude"], 
                       c=dataset["Price in DH"], cmap="viridis", alpha=0.6)
plt.colorbar(scatter, label="Price in DH")
plt.title("Property Distribution by Location")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.tight_layout()
plt.show()
```
