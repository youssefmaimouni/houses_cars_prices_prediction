```python
import pandas as pd
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from sklearn.metrics import r2_score, mean_absolute_percentage_error

# Load the dataset
df = pd.read_csv('Cleaned_dataset.csv')

# Drop unnecessary columns
df = df.drop(columns=['Localisation', 'Latitude', 'Longitude', 'Title', 'Tags', 'Tags_length'])

# Separate features and target
X = df.drop(columns=['Price in DH'])
y = df['Price in DH']

# Define categorical and numerical columns
categorical_cols = ['Type', 'Condition']  # Include 'Type' in categorical columns
numerical_cols = ['Area in m²', 'Rooms', 'Bedrooms', 'Bathrooms', 'Floor', 'House Age in ans']

# Create a ColumnTransformer for preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_cols),  # Scale numerical features
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)  # One-hot encode categorical features
    ])

# Preprocess the data
X_processed = preprocessor.fit_transform(X)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_processed, y, test_size=0.2, random_state=42)

# Define the updated model
model = Sequential()

# Input layer (update input_dim to match the number of features after preprocessing)
model.add(Dense(128, input_dim=X_train.shape[1], activation='relu'))

# Hidden layers
model.add(Dense(64, activation='relu'))
model.add(Dropout(0.2))
model.add(Dense(32, activation='relu'))

# Output layer
model.add(Dense(1, activation='linear'))

# Compile the model
model.compile(optimizer='adam', loss='mean_squared_error')

# Summary of the model
model.summary()

# Train the model
history = model.fit(X_train, y_train, epochs=50, batch_size=32, validation_split=0.2)

# Evaluate the model on the test set
loss = model.evaluate(X_test, y_test)
print(f'Test Loss: {loss}')

# Calculate R² score
y_pred = model.predict(X_test)
r2 = r2_score(y_test, y_pred)
print(f'R² Score: {r2}')

# Calculate MAPE
mape = mean_absolute_percentage_error(y_test, y_pred)
print(f'Mean Absolute Percentage Error (MAPE): {mape:.2f}%')

# Use the model to make predictions on new data
input_data = {
    'Type': ['Appartements'],
    'Area in m²': [120],
    'Rooms': [3],
    'Bedrooms': [2],
    'Bathrooms': [2],
    'Condition': ['Bon état'],
    'Floor': [2],
    'House Age in ans': [10]
}

# Convert the dictionary to a DataFrame
input_df = pd.DataFrame(input_data)

# Preprocess the input data using the same preprocessor
input_processed = preprocessor.transform(input_df)

# Use the trained model to make a prediction
predicted_price = model.predict(input_processed)

# Print the predicted price
print(f'Predicted Price: {predicted_price[0][0]:.2f} DH')
```
